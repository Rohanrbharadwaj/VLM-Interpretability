```python
%%writefile /kaggle/working/jepa/train_projector.py
"""
Stage-1 projector training: frozen I-JEPA (ViT-H/16) + frozen 4-bit Vicuna-7B,
train only the MLP projector. Every checkpoint save also pushes to a persistent
Kaggle Dataset via the Kaggle API, so progress survives session death — set
--kaggle_dataset "username/dataset-slug" to enable this.

NaN recovery (not halt):
- gradient clipping + lower default LR (2e-4) address the root cause directly
- if a batch's loss is NaN/Inf, that batch is skipped (grads discarded) rather
  than applied — training continues to the next batch
- an in-memory backup of the last known-good projector+optimizer state is kept;
  if weights become NaN/Inf despite the above, they're rolled back to that
  backup and training continues from there
- if MAX_CONSECUTIVE_RECOVERIES is hit (default 30), the run does stop — at
  that point every batch is corrupting, which means something is fundamentally
  broken, and looping forever would waste GPU time making zero progress. This
  is presented honestly as a last resort, not the expected path.
"""

import argparse
import os
import json
import shutil
import subprocess
import copy
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig
from PIL import Image

from ijepa_encoder import IJEPAVisionTower


class MLPProjector(nn.Module):
    def __init__(self, vision_hidden_size, llm_hidden_size):
        super().__init__()
        self.proj = nn.Sequential(
            nn.Linear(vision_hidden_size, llm_hidden_size),
            nn.GELU(),
            nn.Linear(llm_hidden_size, llm_hidden_size),
        )

    def forward(self, x):
        return self.proj(x)


class CaptionDataset(Dataset):
    def __init__(self, json_path, image_root):
        with open(json_path, "r") as f:
            self.records = json.load(f)
        self.image_root = image_root

    def __len__(self):
        return len(self.records)

    def __getitem__(self, idx):
        rec = self.records[idx]
        image = Image.open(os.path.join(self.image_root, rec["image"])).convert("RGB")
        return image, rec["caption"]


def collate_fn(batch):
    images, captions = zip(*batch)
    return list(images), list(captions)


def build_training_batch(tokenizer, vision_tower, projector, llm, images, captions, device):
    visual_features = vision_tower(images)
    projected = projector(visual_features)

    tokenized = tokenizer(
        list(captions), return_tensors="pt", padding=True, truncation=True, max_length=128
    ).to(device)
    input_ids = tokenized.input_ids
    attention_mask = tokenized.attention_mask
    labels = input_ids.clone()

    text_embeds = llm.get_input_embeddings()(input_ids)
    inputs_embeds = torch.cat([projected, text_embeds], dim=1)
    model_dtype = next(llm.parameters()).dtype
    inputs_embeds = inputs_embeds.to(model_dtype)

    image_attn = torch.ones(
        (attention_mask.shape[0], projected.shape[1]), device=device, dtype=attention_mask.dtype
    )
    full_attention_mask = torch.cat([image_attn, attention_mask], dim=1)

    image_labels = torch.full(
        (labels.shape[0], projected.shape[1]), -100, device=device, dtype=labels.dtype
    )
    full_labels = torch.cat([image_labels, labels], dim=1)

    return inputs_embeds, full_attention_mask, full_labels


@torch.no_grad()
def run_validation(tokenizer, vision_tower, projector, llm, val_loader, device):
    projector.eval()
    total_loss = 0.0
    n_batches = 0
    for images, captions in val_loader:
        inputs_embeds, attention_mask, labels = build_training_batch(
            tokenizer, vision_tower, projector, llm, images, captions, device
        )
        outputs = llm(inputs_embeds=inputs_embeds, attention_mask=attention_mask, labels=labels)
        total_loss += outputs.loss.item()
        n_batches += 1
    projector.train()
    return total_loss / max(n_batches, 1)


def state_dict_has_nan_or_inf(state_dict):
    for v in state_dict.values():
        if torch.is_tensor(v) and (torch.isnan(v).any() or torch.isinf(v).any()):
            return True
    return False


def push_checkpoint(ckpt_path, kaggle_dataset, step, push_dir="/kaggle/working/ckpt_push"):
    """Copy the checkpoint into push_dir and create a new dataset version via Kaggle API.
    Never raises — a failed push should not kill training."""
    try:
        os.makedirs(push_dir, exist_ok=True)
        meta_path = os.path.join(push_dir, "dataset-metadata.json")
        if not os.path.exists(meta_path):
            with open(meta_path, "w") as f:
                json.dump({"title": "jepa-projector-checkpoints", "id": kaggle_dataset,
                           "licenses": [{"name": "CC0-1.0"}]}, f)
        shutil.copy(ckpt_path, os.path.join(push_dir, "latest_ckpt.pt"))
        with open(os.path.join(push_dir, "STEP.txt"), "w") as f:
            f.write(str(step))
        result = subprocess.run(
            ["kaggle", "datasets", "version", "-p", push_dir, "-m", f"step {step}", "--dir-mode", "zip"],
            check=False, capture_output=True, text=True, timeout=180,
        )
        if result.returncode == 0:
            print(f"[checkpoint push] SUCCESS step {step} -> {kaggle_dataset}")
        else:
            print(f"[checkpoint push] FAILED (exit {result.returncode}) at step {step}: {result.stderr.strip()[:200]}")
    except Exception as e:
        print(f"[checkpoint push] EXCEPTION at step {step}: {e} — continuing training anyway")


def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("--llm_path", type=str, default="lmsys/vicuna-7b-v1.5")
    parser.add_argument("--data_json", type=str, required=True)
    parser.add_argument("--image_root", type=str, required=True)
    parser.add_argument("--val_json", type=str, default=None)
    parser.add_argument("--val_every", type=int, default=50)
    parser.add_argument("--output_dir", type=str, default="/kaggle/working/jepa_projector_ckpt")
    parser.add_argument("--batch_size", type=int, default=2)
    parser.add_argument("--grad_accum_steps", type=int, default=8)
    parser.add_argument("--lr", type=float, default=2e-4)  # lowered from 1e-3 — reduces overflow risk
    parser.add_argument("--grad_clip_norm", type=float, default=1.0)  # caps update size — root-cause fix
    parser.add_argument("--epochs", type=int, default=1)
    parser.add_argument("--save_every", type=int, default=100)
    parser.add_argument("--resume", action="store_true")
    parser.add_argument("--kaggle_dataset", type=str, default=None)
    parser.add_argument("--push_every", type=int, default=200)
    parser.add_argument("--max_consecutive_recoveries", type=int, default=30,
                         help="stop only if this many rollbacks/skips happen in a row with no clean step between")
    args = parser.parse_args()

    device = "cuda" if torch.cuda.is_available() else "cpu"
    os.makedirs(args.output_dir, exist_ok=True)

    vision_tower = IJEPAVisionTower(device=device)
    vision_tower.load_model()

    tokenizer = AutoTokenizer.from_pretrained(args.llm_path)
    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token

    bnb_config = BitsAndBytesConfig(load_in_4bit=True, bnb_4bit_compute_dtype=torch.float16)
    llm = AutoModelForCausalLM.from_pretrained(
        args.llm_path, quantization_config=bnb_config, device_map="auto", low_cpu_mem_usage=True,
    )
    llm.eval()
    for p in llm.parameters():
        p.requires_grad_(False)

    projector = MLPProjector(vision_tower.hidden_size, llm.config.hidden_size).to(device)
    optimizer = torch.optim.AdamW(projector.parameters(), lr=args.lr)

    start_step = 0
    ckpt_path = os.path.join(args.output_dir, "latest_ckpt.pt")
    if args.resume and os.path.exists(ckpt_path):
        ckpt = torch.load(ckpt_path, map_location=device)
        if state_dict_has_nan_or_inf(ckpt["projector"]):
            raise RuntimeError(f"Refusing to resume: {ckpt_path} (step {ckpt.get('step')}) contains NaN/Inf.")
        projector.load_state_dict(ckpt["projector"])
        optimizer.load_state_dict(ckpt["optimizer"])
        start_step = ckpt["step"]
        print(f"Resumed from step {start_step} (verified clean)")

    projector.train()

    dataset = CaptionDataset(args.data_json, args.image_root)
    dataloader = DataLoader(dataset, batch_size=args.batch_size, shuffle=True, collate_fn=collate_fn)

    val_loader = None
    if args.val_json:
        val_dataset = CaptionDataset(args.val_json, args.image_root)
        val_loader = DataLoader(val_dataset, batch_size=args.batch_size, shuffle=False, collate_fn=collate_fn)
        print(f"Validation set loaded: {len(val_dataset)} examples")

    # In-memory rollback target — refreshed every time we confirm a clean state.
    last_good_projector_state = copy.deepcopy(projector.state_dict())
    last_good_optimizer_state = copy.deepcopy(optimizer.state_dict())
    consecutive_recoveries = 0

    step = 0
    for epoch in range(args.epochs):
        for images, captions in dataloader:
            if step < start_step:
                step += 1
                continue

            inputs_embeds, attention_mask, labels = build_training_batch(
                tokenizer, vision_tower, projector, llm, images, captions, device
            )
            outputs = llm(inputs_embeds=inputs_embeds, attention_mask=attention_mask, labels=labels)
            loss = outputs.loss / args.grad_accum_steps

            # --- Skip (don't halt) if this batch's loss itself is already broken ---
            if torch.isnan(loss).any() or torch.isinf(loss).any():
                consecutive_recoveries += 1
                print(f"[nan-recovery] step {step}: NaN/Inf loss — skipping this batch "
                      f"({consecutive_recoveries}/{args.max_consecutive_recoveries} consecutive)")
                optimizer.zero_grad()  # discard any partial grads from this accumulation window
                if consecutive_recoveries >= args.max_consecutive_recoveries:
                    print("!!! Too many consecutive failures — every batch is corrupting. Stopping. !!!")
                    print(f"!!! Last known-good state was step {step - consecutive_recoveries} !!!")
                    torch.save({"projector": last_good_projector_state,
                                "optimizer": last_good_optimizer_state, "step": step}, ckpt_path)
                    return
                step += 1
                continue

            loss.backward()
            torch.nn.utils.clip_grad_norm_(projector.parameters(), max_norm=args.grad_clip_norm)

            if (step + 1) % args.grad_accum_steps == 0:
                optimizer.step()
                optimizer.zero_grad()

                # --- Rollback (don't halt) if weights became corrupted despite the above ---
                if state_dict_has_nan_or_inf(projector.state_dict()):
                    consecutive_recoveries += 1
                    print(f"[nan-recovery] step {step}: weights corrupted after optimizer step — "
                          f"rolling back to last known-good state "
                          f"({consecutive_recoveries}/{args.max_consecutive_recoveries} consecutive)")
                    projector.load_state_dict(last_good_projector_state)
                    optimizer.load_state_dict(last_good_optimizer_state)
                    if consecutive_recoveries >= args.max_consecutive_recoveries:
                        print("!!! Too many consecutive failures — every step is corrupting. Stopping. !!!")
                        torch.save({"projector": last_good_projector_state,
                                    "optimizer": last_good_optimizer_state, "step": step}, ckpt_path)
                        return
                else:
                    # Clean step — this becomes the new rollback target, and the streak resets.
                    consecutive_recoveries = 0
                    last_good_projector_state = copy.deepcopy(projector.state_dict())
                    last_good_optimizer_state = copy.deepcopy(optimizer.state_dict())

            if step % 50 == 0:
                train_loss_val = loss.item() * args.grad_accum_steps
                if val_loader is not None and step % args.val_every == 0:
                    val_loss = run_validation(tokenizer, vision_tower, projector, llm, val_loader, device)
                    print(f"epoch {epoch} step {step} train_loss {train_loss_val:.4f} val_loss {val_loss:.4f}")
                else:
                    print(f"epoch {epoch} step {step} train_loss {train_loss_val:.4f}")

            if step % args.save_every == 0 and step > 0:
                current_state = projector.state_dict()
                if state_dict_has_nan_or_inf(current_state):
                    print(f"[save] step {step}: current state corrupted — saving last known-good instead")
                    torch.save({"projector": last_good_projector_state,
                                "optimizer": last_good_optimizer_state, "step": step}, ckpt_path)
                else:
                    torch.save({"projector": current_state, "optimizer": optimizer.state_dict(), "step": step},
                               ckpt_path)
                if args.kaggle_dataset and step % args.push_every == 0:
                    push_checkpoint(ckpt_path, args.kaggle_dataset, step)

            step += 1

        if val_loader is not None:
            val_loss = run_validation(tokenizer, vision_tower, projector, llm, val_loader, device)
            print(f"=== epoch {epoch} finished — val_loss {val_loss:.4f} ===")
        if args.kaggle_dataset and os.path.exists(ckpt_path):
            push_checkpoint(ckpt_path, args.kaggle_dataset, step)

    final_state = projector.state_dict()
    if state_dict_has_nan_or_inf(final_state):
        final_state = last_good_projector_state
        print("Final state was corrupted — saving last known-good instead.")
    torch.save(final_state, os.path.join(args.output_dir, "mm_projector_final.bin"))
    if args.kaggle_dataset:
        push_checkpoint(os.path.join(args.output_dir, "mm_projector_final.bin"), args.kaggle_dataset, step,
                         push_dir="/kaggle/working/ckpt_push")
    print(f"Done. Final projector saved to {args.output_dir}/mm_projector_final.bin")


if __name__ == "__main__":
    main()
```

```python
%%writefile /kaggle/working/jepa/ijepa_encoder.py
"""
I-JEPA vision tower — v2, simplified via Hugging Face `transformers`.

Supersedes the earlier hand-rolled checkpoint-loading version: I-JEPA is now
officially integrated into `transformers` (IJepaModel/IJepaConfig), so loading
is a one-line `from_pretrained` call instead of manually parsing a .pth.tar
state_dict. Still matches the same minimal interface LLaVA's CLIPVisionTower
exposes (forward -> patch vectors, hidden_size property), so it's a drop-in
swap for clip_encoder.py.

Checkpoint used: facebook/ijepa_vith16_1k (ViT-Huge, patch size 16).
  - Patch 16 (not 14) chosen specifically to keep patch count lower than
    patch-14 variants, given the Kaggle disk-quota constraints discussed
    earlier in caching hidden states.
  - hidden_size for this checkpoint is 1280 — confirm this against
    model.config.hidden_size after loading, since a silent mismatch here
    is exactly the kind of bug that's cheap to catch now and annoying
    to debug after a full projector training run.
  - No ViT-B checkpoint is officially hosted under the `facebook` org on
    Hugging Face (verified directly) — only ViT-H (this one) and ViT-Giant.
    ViT-H is used here specifically because it loads cleanly via
    AutoModel.from_pretrained with no manual state_dict work, which
    matters more for reliability than the extra parameter count costs
    in GPU time, given the encoder is frozen throughout (no gradients,
    so its memory footprint is just weights + one forward pass'
    activations — not weights + gradients + optimizer state).
"""

import torch
import torch.nn as nn
from transformers import AutoModel, AutoImageProcessor


class IJEPAVisionTower(nn.Module):
    def __init__(self, model_id="facebook/ijepa_vith16_1k", device="cuda"):
        super().__init__()
        self.model_id = model_id
        self.device_ = device
        self.is_loaded = False

    def load_model(self):
        if self.is_loaded:
            return

        self.image_processor = AutoImageProcessor.from_pretrained(self.model_id)
        self.vision_tower = AutoModel.from_pretrained(self.model_id)

        # Frozen throughout — this is the control experiment's whole point:
        # Vicuna must stay identical between the CLIP and JEPA versions, and
        # the encoder itself is never fine-tuned in either version either.
        self.vision_tower.requires_grad_(False)
        self.vision_tower.eval()
        self.vision_tower.to(self.device_)

        print(f"[IJEPAVisionTower] loaded {self.model_id}")
        print(f"[IJEPAVisionTower] hidden_size = {self.hidden_size}")
        print(f"[IJEPAVisionTower] num_patches = {self.num_patches}")

        self.is_loaded = True

    @torch.no_grad()
    def forward(self, pil_images):
        """
        pil_images: a single PIL Image or a list of PIL Images.
        Returns: (B, num_patches, hidden_size) tensor of patch vectors —
        this is IJepaModel's last_hidden_state, with no CLS token, no
        pooling — exactly the sequence-of-patch-vectors shape LLaVA's
        projector expects as input.
        """
        if not isinstance(pil_images, list):
            pil_images = [pil_images]

        inputs = self.image_processor(pil_images, return_tensors="pt").to(self.device_)
        outputs = self.vision_tower(**inputs)
        return outputs.last_hidden_state

    @property
    def hidden_size(self):
        return self.vision_tower.config.hidden_size

    @property
    def num_patches(self):
        img_size = self.vision_tower.config.image_size
        patch_size = self.vision_tower.config.patch_size
        if isinstance(img_size, int):
            img_size = (img_size, img_size)
        if isinstance(patch_size, int):
            patch_size = (patch_size, patch_size)
        return (img_size[0] // patch_size[0]) * (img_size[1] // patch_size[1])

    @property
    def dtype(self):
        return next(self.vision_tower.parameters()).dtype
```

```python
import os, json, torch, shutil, subprocess, zipfile

DATASET = "ponishbhatia/jepa-projector-checkpoints"
PULL_DIR = "/kaggle/working/ckpt_pull"
OUTPUT_DIR = "/kaggle/working/jepa_projector_ckpt"
MAX_VERSION_TO_SCAN = 23

def state_dict_has_nan_or_inf(state_dict):
    for v in state_dict.values():
        if torch.isnan(v).any() or torch.isinf(v).any():
            return True
    return False

def try_version(version_number):
    if os.path.exists(PULL_DIR):
        shutil.rmtree(PULL_DIR)
    os.makedirs(PULL_DIR, exist_ok=True)

    zip_path = os.path.join(PULL_DIR, "download.zip")
    url = f"https://www.kaggle.com/api/v1/datasets/download/{DATASET}?datasetVersionNumber={version_number}"

    result = subprocess.run(
        ["curl", "-sL", "-u", f"{os.environ['KAGGLE_USERNAME']}:{os.environ['KAGGLE_KEY']}",
         "-o", zip_path, url],
        capture_output=True, text=True, timeout=120,
    )

    if not os.path.exists(zip_path) or os.path.getsize(zip_path) < 1000:
        return None

    try:
        with zipfile.ZipFile(zip_path, "r") as z:
            z.extractall(PULL_DIR)
    except zipfile.BadZipFile:
        # got an HTML error page instead of a real zip — print it once for diagnosis
        with open(zip_path, "r", errors="ignore") as f:
            content = f.read(500)
        print(f"  version {version_number}: not a valid zip — response: {content[:200]}")
        return None

    ckpt_path = os.path.join(PULL_DIR, "latest_ckpt.pt")
    if not os.path.exists(ckpt_path):
        return None

    step = None
    step_path = os.path.join(PULL_DIR, "STEP.txt")
    if os.path.exists(step_path):
        with open(step_path) as f:
            step = f.read().strip()

    ckpt = torch.load(ckpt_path, map_location="cpu")
    corrupted = state_dict_has_nan_or_inf(ckpt["projector"])
    return {"version": version_number, "step": step, "corrupted": corrupted, "ckpt_path": ckpt_path}


print(f"Scanning {DATASET} versions (newest first)...\n")
found_clean = None
for v in range(MAX_VERSION_TO_SCAN, 0, -1):
    info = try_version(v)
    if info is None:
        continue
    status = "CORRUPTED" if info["corrupted"] else "CLEAN"
    print(f"version {info['version']}: step {info['step']} — {status}")
    if not info["corrupted"] and found_clean is None:
        found_clean = info

print()
if found_clean:
    print(f"✅ Using version {found_clean['version']} (step {found_clean['step']}) — confirmed clean")
    os.makedirs(OUTPUT_DIR, exist_ok=True)
    dest = os.path.join(OUTPUT_DIR, "latest_ckpt.pt")
    shutil.copy(found_clean["ckpt_path"], dest)
    print(f"Copied to {dest} — ready to resume with --resume")
else:
    print("❌ No clean checkpoint found in the scanned range.")
```
