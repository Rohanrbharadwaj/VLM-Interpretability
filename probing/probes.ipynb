```python
import os
import json
import numpy as np
import pandas as pd
import torch
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import balanced_accuracy_score, roc_auc_score

INPUT = "/kaggle/input/datasets/nocturnalnerd18"
TYPES = ["attribute", "object", "relation"]
USERNAME = "nocturnalnerd18"
NUM_LAYERS = 33

with open(f"{INPUT}/vlm-labeled-examples/labeled_examples.json") as f:
    labels = pd.DataFrame(json.load(f))
labels = labels[labels["cache"] == "question_aware"].copy()
labels["truth"] = labels["ground_truth_answer"].str.strip().str.lower()
labels["is_yes_truth"] = (labels["truth"] == "yes").astype(int)
labels["false_yes"] = labels["false_yes"].astype(int)
print(len(labels), "labeled rows")
```

```python
type_data = {}
for t in TYPES:
    with open(f"{INPUT}/probe-features-{t}/question_ids.json") as f:
        qids = json.load(f)
    p1 = torch.load(f"{INPUT}/probe-features-{t}/p1.pt", mmap=True)
    p2 = torch.load(f"{INPUT}/probe-features-{t}/p2.pt", mmap=True)
    type_data[t] = {"row_of": pd.Series(range(len(qids)), index=qids), "p1": p1, "p2": p2}
    print(f"{t:10s} {p1.shape}")
```

```python
GROUPS = [
    ("attribute", None, "attribute"),
    ("object", None, "object"),
    ("relation", None, "relation_all"),
    ("relation", "amber", "relation_amber"),
    ("relation", "reefknot", "relation_reefknot"),
]


def group_frame(t, source_filter):
    df = labels[labels["hallucination_type"] == t].copy()
    if source_filter is not None:
        df = df[df["source"] == source_filter]
    df["row"] = df["question_id"].map(type_data[t]["row_of"])
    assert df["row"].notna().all()
    df["row"] = df["row"].astype(int)
    return df.reset_index(drop=True)
```

```python
import warnings
from sklearn.exceptions import ConvergenceWarning


def _fit_once(Xtr_s, ytr, C, max_iter):
    with warnings.catch_warnings():
        warnings.simplefilter("ignore", category=ConvergenceWarning)
        clf = LogisticRegression(class_weight="balanced", C=C, max_iter=max_iter)
        clf.fit(Xtr_s, ytr)
    return clf, int(clf.n_iter_[0]) < max_iter


def fit_eval_layer(Xtr, ytr, Xval, yval, Xte, yte, C=0.01, max_iter=3000):
    if ytr.nunique() < 2:
        return None
    scaler = StandardScaler().fit(Xtr)
    Xtr_s = scaler.transform(Xtr)

    clf, converged = _fit_once(Xtr_s, ytr, C, max_iter)
    if not converged:
        clf, converged = _fit_once(Xtr_s, ytr, C / 10, max_iter * 2)

    out = {"converged": converged}
    for name, X, y in [("val", Xval, yval), ("test", Xte, yte)]:
        if len(y) == 0 or y.nunique() < 2:
            out[f"{name}_bal_acc"], out[f"{name}_auroc"] = None, None
            continue
        Xs = scaler.transform(X)
        pred, proba = clf.predict(Xs), clf.predict_proba(Xs)[:, 1]
        out[f"{name}_bal_acc"] = round(float(balanced_accuracy_score(y, pred)), 4)
        out[f"{name}_auroc"] = round(float(roc_auc_score(y, proba)), 4)
    return out
```

```python
sweep_rows, picked = [], []

for t, src, group_name in GROUPS:
    df = group_frame(t, src)
    for feat in ["p1", "p2"]:
        X_all = type_data[t][feat][df["row"].to_numpy()].float().numpy()

        for target, y_col in [("A", "is_yes_truth"), ("B", "false_yes")]:
            if target == "A":
                sub_df, X_sub = df, X_all
            else:
                mask = (df["model_answer"] == "yes").to_numpy()
                sub_df, X_sub = df[mask].reset_index(drop=True), X_all[mask]

            tr = (sub_df.split == "train").to_numpy()
            va = (sub_df.split == "val").to_numpy()
            te = (sub_df.split == "test").to_numpy()
            ytr, yval, yte = sub_df.loc[tr, y_col], sub_df.loc[va, y_col], sub_df.loc[te, y_col]
            if ytr.nunique() < 2 or va.sum() == 0 or te.sum() == 0:
                print(f"skip {group_name} {target} {feat}: insufficient data")
                continue

            best = None
            for layer in range(NUM_LAYERS):
                res = fit_eval_layer(X_sub[tr, layer, :], ytr, X_sub[va, layer, :], yval, X_sub[te, layer, :], yte)
                if res is None:
                    continue
                row = {"group": group_name, "target": target, "feature": feat, "layer": layer, "n_train": int(tr.sum()), **res}
                sweep_rows.append(row)
                if best is None or (res["val_bal_acc"] or -1) > (best["val_bal_acc"] or -1):
                    best = row
            if best:
                picked.append(best)
                flag = "" if best["converged"] else "  [NOT CONVERGED]"
                print(f"{group_name:18s} {target} {feat}: best layer {best['layer']:2d} | "
                      f"val {best['val_bal_acc']} | test {best['test_bal_acc']} (auroc {best['test_auroc']}){flag}")
        del X_all

sweep_df = pd.DataFrame(sweep_rows)
picked_df = pd.DataFrame(picked)
```

```python
print("layer-0 sanity, p1 should sit at the group's chance/majority rate:")
print(sweep_df[(sweep_df.feature == "p1") & (sweep_df.layer == 0)][["group", "target", "test_bal_acc", "test_auroc", "converged"]].to_string(index=False))

n_bad = (~sweep_df["converged"]).sum()
print(f"\n{n_bad} of {len(sweep_df)} fits still not converged after the fallback")
if n_bad:
    print(sweep_df[~sweep_df["converged"]][["group", "target", "feature", "layer"]].to_string(index=False))
```

```python
OUT = "/kaggle/working/probe_results"
os.makedirs(OUT, exist_ok=True)

sweep_df.to_json(f"{OUT}/layer_sweep.json", orient="records")
picked_df.to_json(f"{OUT}/best_layer_summary.json", orient="records")

slug = "probe-results"
with open(f"{OUT}/dataset-metadata.json", "w") as f:
    json.dump({"title": slug, "id": f"{USERNAME}/{slug}", "licenses": [{"name": "CC0-1.0"}]}, f)

!kaggle datasets version -p {OUT} -m "rerun with converged fits (C=0.01, automatic fallback retry, 0/660 unconverged)"
```
