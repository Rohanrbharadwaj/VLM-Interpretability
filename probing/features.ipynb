```python
import os
import re
import glob
import json
import torch

INPUT = "/kaggle/input/datasets/nocturnalnerd18"
TYPES = ["attribute", "object", "relation"]
USERNAME = "nocturnalnerd18"
EXPECTED_ROWS = {"attribute": 7628, "object": 3000, "relation": 6506}
```

```python
def batch_number(path):
    return int(re.search(r"_batch_(\d+)\.pt$", path).group(1))


def extract_type(t):
    files = sorted(glob.glob(f"{INPUT}/question-aware-cache-{t}/{t}_batch_*.pt"), key=batch_number)
    question_ids, p1_list, p2_list = [], [], []
    image_features = {}
    for path in files:
        batch = torch.load(path, mmap=True)
        for e in batch:
            question_ids.append(e["question_id"])
            p1_list.append(e["p1_hidden"])
            p2_list.append(e["p2_hidden"])
            if e["image_id"] not in image_features:
                image_features[e["image_id"]] = {
                    "pooled_image_hidden": e["pooled_image_hidden"],
                    "pooled_encoder_embedding": e["pooled_encoder_embedding"][0],
                }
        del batch
    p1 = torch.stack(p1_list)
    p2 = torch.stack(p2_list)
    return question_ids, p1, p2, image_features
```

```python
OUT = "/kaggle/working/probe_features"

for t in TYPES:
    t_dir = f"{OUT}/{t}"
    os.makedirs(t_dir, exist_ok=True)

    question_ids, p1, p2, image_features = extract_type(t)
    assert len(question_ids) == EXPECTED_ROWS[t], (t, len(question_ids))
    assert len(set(question_ids)) == len(question_ids)
    assert p1.shape == (len(question_ids), 33, 4096), p1.shape
    assert p2.shape == p1.shape
    assert not torch.isnan(p1).any() and not torch.isnan(p2).any()
    assert not torch.isnan(torch.stack([v["pooled_image_hidden"] for v in image_features.values()])).any()

    layer0 = p1[:, 0, :]
    print(f"{t:10s} rows {len(question_ids)} | unique images {len(image_features)} | "
          f"layer0 identical within type {bool((layer0 == layer0[0]).all())}")

    torch.save(p1, f"{t_dir}/p1.pt")
    torch.save(p2, f"{t_dir}/p2.pt")
    torch.save(image_features, f"{t_dir}/image_features.pt")
    with open(f"{t_dir}/question_ids.json", "w") as f:
        json.dump(question_ids, f)

    del p1, p2, image_features
```

```python
layer0_by_type = {}
for t in TYPES:
    p1 = torch.load(f"{OUT}/{t}/p1.pt", mmap=True)
    layer0_by_type[t] = p1[0, 0, :].clone()
    del p1

ref = layer0_by_type["attribute"]
for t in TYPES:
    print(f"{t:10s} layer0 matches attribute: {bool(torch.equal(layer0_by_type[t], ref))}")
```
