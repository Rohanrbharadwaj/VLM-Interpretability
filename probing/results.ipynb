```python
import os
import time
import json
import warnings
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import balanced_accuracy_score, roc_auc_score
from sklearn.exceptions import ConvergenceWarning

INPUT = "/kaggle/input/datasets/nocturnalnerd18"
TYPES = ["attribute", "object", "relation"]
USERNAME = "nocturnalnerd18"
NUM_LAYERS = 33
N_BOOT = 500

with open(f"{INPUT}/vlm-labeled-examples/labeled_examples.json") as f:
    labels = pd.DataFrame(json.load(f))
labels = labels[labels["cache"] == "question_aware"].copy()
labels["false_yes"] = labels["false_yes"].astype(int)

with open(f"{INPUT}/probe-baselines/baseline_results.json") as f:
    baseline_df = pd.DataFrame(json.load(f))

type_data = {}
for t in TYPES:
    with open(f"{INPUT}/probe-features-{t}/question_ids.json") as f:
        qids = json.load(f)
    p1 = torch.load(f"{INPUT}/probe-features-{t}/p1.pt", mmap=True)
    p2 = torch.load(f"{INPUT}/probe-features-{t}/p2.pt", mmap=True)
    type_data[t] = {"row_of": pd.Series(range(len(qids)), index=qids), "p1": p1, "p2": p2}

GROUPS = [
    ("attribute", None, "attribute"),
    ("object", None, "object"),
    ("relation", None, "relation_all"),
    ("relation", "amber", "relation_amber"),
    ("relation", "reefknot", "relation_reefknot"),
]


def group_frame(t, source_filter):
    df = labels[labels["hallucination_type"] == t].copy()
    if source_filter is not None:
        df = df[df["source"] == source_filter]
    df["row"] = df["question_id"].map(type_data[t]["row_of"])
    assert df["row"].notna().all()
    df["row"] = df["row"].astype(int)
    return df.reset_index(drop=True)
```

```python
def _fit_once(Xtr_s, ytr, C, max_iter):
    with warnings.catch_warnings():
        warnings.simplefilter("ignore", category=ConvergenceWarning)
        clf = LogisticRegression(class_weight="balanced", C=C, max_iter=max_iter)
        clf.fit(Xtr_s, ytr)
    return clf, int(clf.n_iter_[0]) < max_iter


def fit_layer(Xtr, ytr, C=0.01, max_iter=3000):
    scaler = StandardScaler().fit(Xtr)
    Xtr_s = scaler.transform(Xtr)
    clf, converged = _fit_once(Xtr_s, ytr, C, max_iter)
    if not converged:
        clf, converged = _fit_once(Xtr_s, ytr, C / 10, max_iter * 2)
    return clf, scaler, converged


def bootstrap_ci(y, pred, proba, image_ids, n_boot, rng):
    image_to_idx = {}
    for i, im in enumerate(image_ids):
        image_to_idx.setdefault(im, []).append(i)
    unique_images = np.array(list(image_to_idx.keys()))

    boot_bal, boot_auc = [], []
    for _ in range(n_boot):
        sampled = rng.choice(unique_images, size=len(unique_images), replace=True)
        idx = np.concatenate([image_to_idx[im] for im in sampled])
        y_b, pr_b, p_b = y[idx], pred[idx], proba[idx]
        if len(np.unique(y_b)) < 2:
            continue
        boot_bal.append(balanced_accuracy_score(y_b, pr_b))
        boot_auc.append(roc_auc_score(y_b, p_b))
    lo_hi = lambda arr: (float(np.percentile(arr, 2.5)), float(np.percentile(arr, 97.5))) if arr else (None, None)
    bal_lo, bal_hi = lo_hi(boot_bal)
    auc_lo, auc_hi = lo_hi(boot_auc)
    return bal_lo, bal_hi, auc_lo, auc_hi
```

```python
rng = np.random.default_rng(123)
curve_rows = []

for t, src, group_name in GROUPS:
    df = group_frame(t, src)
    mask = (df["model_answer"] == "yes").to_numpy()
    sub_df = df[mask].reset_index(drop=True)
    tr = (sub_df.split == "train").to_numpy()
    va = (sub_df.split == "val").to_numpy()
    te = (sub_df.split == "test").to_numpy()
    ytr, yte = sub_df.loc[tr, "false_yes"].to_numpy(), sub_df.loc[te, "false_yes"].to_numpy()
    test_images = sub_df.loc[te, "image_id"].to_numpy()

    if len(np.unique(ytr)) < 2 or te.sum() == 0:
        print(f"skip {group_name}: insufficient data")
        continue

    for feat in ["p1", "p2"]:
        X_all = type_data[t][feat][df["row"].to_numpy()].float().numpy()
        X_sub = X_all[mask]
        t0 = time.time()
        for layer in range(NUM_LAYERS):
            clf, scaler, converged = fit_layer(X_sub[tr, layer, :], ytr)
            proba = clf.predict_proba(scaler.transform(X_sub[te, layer, :]))[:, 1]
            pred = clf.predict(scaler.transform(X_sub[te, layer, :]))
            point_bal = balanced_accuracy_score(yte, pred)
            point_auc = roc_auc_score(yte, proba)
            bal_lo, bal_hi, auc_lo, auc_hi = bootstrap_ci(yte, pred, proba, test_images, N_BOOT, rng)
            curve_rows.append({
                "group": group_name, "feature": feat, "layer": layer,
                "bal_acc": round(float(point_bal), 4), "auroc": round(float(point_auc), 4),
                "bal_acc_lo": round(bal_lo, 4) if bal_lo is not None else None,
                "bal_acc_hi": round(bal_hi, 4) if bal_hi is not None else None,
                "auroc_lo": round(auc_lo, 4) if auc_lo is not None else None,
                "auroc_hi": round(auc_hi, 4) if auc_hi is not None else None,
                "converged": converged,
            })
        del X_all
        print(f"{group_name:18s} {feat}: done in {time.time()-t0:.0f}s")

curve_df = pd.DataFrame(curve_rows)
```

```python
PLOT_DIR = "/kaggle/working/plots"
os.makedirs(PLOT_DIR, exist_ok=True)
COLORS = {"p1": "#1f77b4", "p2": "#d62728"}
BASELINE_STYLES = {"text": ("gray", "--"), "vision_encoder": ("green", ":"), "logit_gap": ("purple", "-.")}

for _, _, group_name in GROUPS:
    fig, ax = plt.subplots(figsize=(8, 5))
    for feat in ["p1", "p2"]:
        rows = curve_df[(curve_df.group == group_name) & (curve_df.feature == feat)].sort_values("layer")
        if rows.empty:
            continue
        ax.plot(rows["layer"], rows["auroc"], label=feat, color=COLORS[feat])
        ax.fill_between(rows["layer"], rows["auroc_lo"], rows["auroc_hi"], color=COLORS[feat], alpha=0.15)

    base_rows = baseline_df[(baseline_df.group == group_name) & (baseline_df.target == "B")]
    for name, (color, style) in BASELINE_STYLES.items():
        vals = base_rows[base_rows.baseline == name]["test_auroc"].values
        if len(vals) and vals[0] is not None:
            ax.axhline(vals[0], color=color, linestyle=style, linewidth=1, label=f"{name} baseline")

    ax.set_xlabel("Layer")
    ax.set_ylabel("Test AUROC (false-yes vs true-yes)")
    ax.set_title(group_name)
    ax.set_ylim(0.3, 1.0)
    ax.legend(fontsize=8)
    fig.tight_layout()
    fig.savefig(f"{PLOT_DIR}/{group_name}_layerwise.png", dpi=150)
    plt.show()
    plt.close(fig)
```

```python
OUT = "/kaggle/working/probe_layer_curves"
os.makedirs(OUT, exist_ok=True)

curve_df.to_json(f"{OUT}/layer_curve_with_ci.json", orient="records")
for f in os.listdir(PLOT_DIR):
    import shutil
    shutil.copy(f"{PLOT_DIR}/{f}", f"{OUT}/{f}")

slug = "probe-layer-curves"
with open(f"{OUT}/dataset-metadata.json", "w") as f:
    json.dump({"title": slug, "id": f"{USERNAME}/{slug}", "licenses": [{"name": "CC0-1.0"}]}, f)

!kaggle datasets create -p {OUT}
```
